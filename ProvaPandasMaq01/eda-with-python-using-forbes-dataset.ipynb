# Gender analysis

Vamos analisar o gênero dos bilionários com o método `value_counts`.

```python
# Use o campo 'gender' e verifique o metodo value_counts
df['gender'].value_counts
```

A maioria dos bilionários são homens. Vamos analisar a porcentagem de homens e mulheres com o parâmetro `normalize=True` ativado.

```python
# o metodo value_counts possui  um parametro normalize=True
# mostre a porcentagem do generos e não quantidade
df['gender'].value_counts(normalize=True)
```

Como você pode ver, 89% dos bilionários são homens e 11% são mulheres. <br/> 
Vamos analisar as médias de idade por gênero. Para isso, vou usar o método `groupby`.

```python
#Agrupe o mesmo e drie um novo daframe chamdado  df_gender
df_gender = df.groupby("gender")["age"]
```

Vamos calcular as idades médias com o método `mean`.

```python
# use o metodo mean nas idade neste novo df (df_gender)
df_gender.head()
df_gender.mean()
```

Como se pode observar, a idade média dos homens é de 64 anos e a das mulheres, de 63 anos.
