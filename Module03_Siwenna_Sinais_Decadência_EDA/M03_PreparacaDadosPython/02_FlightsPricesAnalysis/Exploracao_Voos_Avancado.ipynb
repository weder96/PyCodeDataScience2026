**RECURSOS**
Os vários recursos do conjunto de dados limpo são explicados abaixo:

1) Airline: O nome da companhia aérea é armazenado na coluna "airline". 

2) Flight: O voo armazena informações sobre o código do voo da aeronave. É uma característica categórica.

3) Source City: Cidade de onde o voo parte. É uma característica categórica com 6 cidades únicas.

4) Departure Time: Esta é uma característica categórica derivada, criada agrupando períodos de tempo em intervalos. Ela armazena informações sobre o horário de partida e possui 6 rótulos de horário únicos.

5) Stops: Uma característica categórica com 3 valores distintos que armazena o número de escalas entre as cidades de origem e destino.

6) Arrival Time: Esta é uma característica categórica derivada, criada agrupando intervalos de tempo em intervalos. Ela possui seis rótulos de horário distintos e armazena informações sobre o horário de chegada.

7) Destination City: Cidade onde o voo pousará. É uma característica categórica com 6 cidades únicas.

8) Class: Uma característica categórica que contém informações sobre a classe do assento; Possui dois valores distintos: Negócios e Economia.

9) Duration: Uma variável contínua que exibe o tempo total de viagem entre cidades, em horas.

10) Days Left: Esta é uma característica derivada, calculada subtraindo-se a data da reserva da data da viagem.

11) Price: A variável alvo armazena informações sobre o preço da passagem.
