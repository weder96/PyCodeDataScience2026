### Missão 1: Importação e Setup
**Tarefa:** Importe as bibliotecas pandas e numpy.

```python
import pandas as pd
import numpy as np
```

### Missão 2: Ingestão de Dados Brutos
**Tarefa:** Leia o arquivo 'flights_data.csv' e armazene-o em uma variável chamada `df`. Exiba as 3 primeiras linhas.

```python
# Carregando o dataset e visualizando as primeiras linhas
df = pd.read_csv('flights_data.csv')
df.head(3)
```

### Missão 3: Auditoria de Qualidade (Data Quality)
**Tarefa:** Use um método do Pandas para exibir as informações do dataset (tipos de colunas e contagem de não-nulos).

```python
# Exibe tipos de dados, quantidade de registros e valores nulos
df.info()
```

### Missão 4: Mapeamento de Concorrência
**Tarefa:** Liste todos os valores únicos presentes na coluna `airline`.

```python
# Lista as companhias aéreas únicas
df['airline'].unique()
```

### Missão 5: Logística de Hubs (Frequência)
**Tarefa:** Conte o número de voos que partem de cada cidade (coluna `source_city`).

```python
# Conta a frequência de cada cidade de origem
df['source_city'].value_counts()
```

### Missão 6: SLA de Atendimento (Filtro Simples)
**Tarefa:** Filtre o DataFrame para exibir apenas os voos operados pela 'Vistara'.

```python
# Filtro simples por igualdade
df_vistara = df[df['airline'] == 'Vistara']
df_vistara.head()
```

### Missão 7: Filtro Multicritério para Agência Low-Cost
**Tarefa:** Filtre os voos da classe 'Economy', com partida no período 'Morning' e que custem menos de 5000 rupias.

```python
# Filtro com múltiplos critérios usando o operador & (AND)
df_promo = df[(df['class'] == 'Economy') & (df['departure_time'] == 'Morning') & (df['price'] < 5000)]
df_promo.head()
```

### Missão 8: Ofertas Relâmpago (Ordenação)
**Tarefa:** Ordene o DataFrame pelo preço (`price`) de forma crescente e exiba as 10 passagens mais baratas.

```python
# Ordenação crescente (ascending=True é o padrão)
df.sort_values(by='price').head(10)
```

### Missão 9: Análise de Precificação por Segmento
**Tarefa:** Calcule a média e a mediana de preço (`price`) agrupado pela classe do voo (`class`).

```python
# Agrupamento simples com múltiplas métricas agregadas
df.groupby('class')['price'].agg(['mean', 'median'])
```

### Missão 10: Otimização de Malha Aérea
**Tarefa:** Agrupe pelos campos `source_city` e `destination_city` e calcule a duração média (`duration`) de cada rota.

```python
# Agrupamento por múltiplas colunas (lista)
df.groupby(['source_city', 'destination_city'])['duration'].mean().reset_index()
```

### Missão 11: Previsão de Faturamento Bruto
**Tarefa:** Agrupe pela companhia (`airline`) e calcule a soma total (faturamento bruto potencial) da coluna de preços.

```python
# Somatório agrupado por categoria
df.groupby('airline')['price'].sum().sort_values(ascending=False)
```

### Missão 12: Internacionalização de Receita (Math)
**Tarefa:** Crie uma nova coluna chamada `price_usd` dividindo o preço atual por 83 (taxa de conversão simulada).

```python
# Operação matemática em série (vetorizada)
df['price_usd'] = df['price'] / 83
df[['price', 'price_usd']].head()
```

### Missão 13: Planejamento de Catering com Regras (np.where)
**Tarefa:** Crie uma coluna chamada `flight_type` usando `np.where`: se a duração for maior que 2.5, classifique como 'Longo', caso contrário, 'Curto'.

```python
# Condicional vetorizada usando numpy
df['flight_type'] = np.where(df['duration'] > 2.5, 'Longo', 'Curto')
df[['duration', 'flight_type']].head()
```

### Missão 14: Padronização para Sistemas Legados (Strings)
**Tarefa:** Converta todos os valores da coluna `destination_city` para letras maiúsculas.

```python
# Manipulação de strings usando o acessor .str
df['destination_city'] = df['destination_city'].str.upper()
df['destination_city'].head()
```

### Missão 15: Pipeline Robusto contra Falhas (Nulls)
**Tarefa:** Remova qualquer linha do DataFrame que contenha algum valor nulo (use dropna).

```python
# Remove linhas com qualquer valor NaN
df = df.dropna()
# Conferindo se sobrou algum nulo
print('Total nulos:', df.isna().sum().sum())
```

### Missão 16: Campanha de Push Notification de Última Hora
**Tarefa:** Filtre e exiba os voos onde os dias restantes (`days_left`) sejam menores ou iguais a 3.

```python
# Filtro matemático simples
df[df['days_left'] <= 3].head()
```

### Missão 17: Relatório Executivo Condensado (.agg)
**Tarefa:** Use `.groupby()` combinado com o método `.agg()` para calcular preço mínimo, máximo e médio, quebrado por companhia aérea.

```python
# Agregação múltipla em uma única tabela
df.groupby('airline')['price'].agg(['min', 'max', 'mean'])
```

### Missão 18: Mapeamento de Conexões de Frota (Crosstab)
**Tarefa:** Crie uma tabela de contingência (cross-tabulation) cruzando as colunas `airline` e `stops`.

```python
# Criação de matriz de frequências
pd.crosstab(df['airline'], df['stops'])
```

### Missão 19: Detecção de Anomalias Premium (Percentil)
**Tarefa:** Utilize uma função do numpy (ex: `np.percentile`) para encontrar o preço correspondente ao percentil 95 do dataset.

```python
# Calculando percentis com numpy
p95 = np.percentile(df['price'], 95)
print(f'Percentil 95 do preço: {p95:.2f}')
```

### Missão 20: Custo Operacional de Adicional Noturno
**Tarefa:** Utilizando operadores booleanos lógicos, conte quantos voos partem à noite ('Night') OU chegam à noite ('Night').

```python
# Filtro com operador | (OR) e função len() para contar
voos_noturnos = df[(df['departure_time'] == 'Night') | (df['arrival_time'] == 'Night')]
print(f'Total de voos noturnos: {len(voos_noturnos)}')
```

### Missão 21: Compliance e Proteção de Dados (Drop)
**Tarefa:** Faça o drop (remoção) da coluna `flight` do DataFrame.

```python
# Remoção de colunas
df = df.drop(columns=['flight'])
df.columns
```

### Missão 22: Governança de Metadados (Rename)
**Tarefa:** Renomeie a coluna `days_left` para `dias_para_partida` e `duration` para `duracao_horas`.

```python
# Renomeando colunas através de um dicionário
df = df.rename(columns={'days_left': 'dias_para_partida', 'duration': 'duracao_horas'})
df.columns
```

### Missão 23: Rotas Executivas VIP (Filtro Complexo)
**Tarefa:** Filtre os voos da classe 'Business' saindo de 'Delhi' para 'Mumbai' que possuam 'zero' stops.

```python
# Filtro avançado
df_vip = df[(df['class'] == 'Business') & (df['source_city'] == 'Delhi') & (df['destination_city'] == 'MUMBAI') & (df['stops'] == 'zero')]
df_vip.head()
```

### Missão 24: Taxa Dinâmica de Emissão de Carbono (Apply)
**Tarefa:** Use uma função lambda com `.apply()` na coluna de preço para aumentar em 5% o valor das passagens APENAS se a companhia for 'Air India'.

```python
# Aplicação de função personalizada linha a linha
df['price'] = df.apply(lambda row: row['price'] * 1.05 if row['airline'] == 'Air India' else row['price'], axis=1)
# Exibindo resultado de Air India vs outras
df.groupby('airline')['price'].mean()
```

### Missão 25: Análise de Tendências Históricas (Simulação de Anos)
**Tarefa:** Use `np.random.choice([2023, 2024, 2025])` para gerar dados e criar uma nova coluna chamada `ano_simulado`. Em seguida, agrupe os dados por esta nova coluna para contar.

```python
# Gerando dados sintéticos (mock data) baseados no tamanho do dataframe (len)
df['ano_simulado'] = np.random.choice([2023, 2024, 2025], size=len(df))
# Contagem agrupada
df.groupby('ano_simulado').size()
```
