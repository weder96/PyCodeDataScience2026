### Learning 8: Padronização de Sistemas Legados

**Tarefa 1:** Como o nosso dataset possui a coluna Region (Região), podemos usar uma regra de negócio para definir que cada região operacional representa uma filial diferente.Com base na distribuição dos nossos dados, vamos mapear as filiais assim:Região West (Oeste) $\rightarrow$ Filial ARegião East (Leste) $\rightarrow$ Filial BRegião Central (Central) $\rightarrow$ Filial CRegião South (Sul) $\rightarrow$ Filial D

**Tarefa 2:** Renomeie as colunas 'Branch' para 'Filial' e 'City' para 'Cidade'.

```python

# 2. Cria um dicionário com a regra de mapeamento
regras_filiais = {
    'West': 'A',
    'East': 'B',
    'Central': 'C',
    'South': 'D'
}

# 3. Cria a nova coluna 'Branch' usando a função .map()
df['Branch'] = df['Region'].map(regras_filiais)

# 4. Verifica o resultado contando quantas linhas cada filial possui
print(df['Branch'].value_counts())
# Renomeia colunas usando um dicionário
df = df.rename(columns={'Branch': 'Filial', 'City': 'Cidade'})
df.head(2)
```

### Learning 12:Rastreamento Específico (Filtro AND)
**Tarefa:** Filtre o dataset supermarket.csv para exibir apenas os pedidos que foram enviados por entrega rápida (Ship Mode igual a 'First Class') E que pertençam à categoria de tecnologia (Category igual a 'Technology').

```python
# Aplica o filtro com duas condições usando o operador '&' (AND)
# IMPORTANTE: No Pandas, cada condição deve estar entre parênteses
df_filtrado = df[(df['Ship Mode'] == 'First Class') & (df['Category'] == 'Technology')]
# Exibe a quantidade de registros que atendem aos dois critérios
print(f"Total de pedidos de Tecnologia com entrega First Class: {len(df_filtrado)}")
```

### Learning 13:  Foco em Mobiliário (Filtro OR)
Tarefa: Filtre o dataset supermarket.csv para exibir apenas as vendas que pertençam às subcategorias de cadeiras ('Chairs') OU mesas ('Tables').

```python
# Método 1: Usando o operador lógico '|' (OR)
df_moveis = df[(df['Sub-Category'] == 'Chairs') | (df['Sub-Category'] == 'Tables')]

# Método 2 (Melhor prática): Usando .isin() para múltiplos valores na mesma coluna
df_moveis_isin = df[df['Sub-Category'].isin(['Chairs', 'Tables'])]

# Exibe o resultado (ambos os métodos geram o mesmo DataFrame)
print(f"Total de vendas de Cadeiras e Mesas: {len(df_moveis)}")
```

### Learning 24: Matriz Executiva de Avaliações (Pivot Table)
**Tarefa:** Construa uma Pivot Table calculando a média de 'Sales' por 'Cidade' e 'Filial'.

```python
# 2. Recria a coluna 'Branch' (Filial) baseada nas Regiões (conforme Learning 15)
df['Filial'] = df['Region'].map({'West': 'A', 'East': 'B', 'Central': 'C', 'South': 'D'})

# 3. Cria a Tabela Dinâmica usando Sales (Vendas), City (Cidade) e Branch (Filial)
tabela_dinamica = pd.pivot_table(df, values='Sales', index='Cidade', columns='Filial', aggfunc='mean')

# 4. Formata o resultado para ficar legível (arredonda para 2 casas e substitui 'NaN' por '-')
print(tabela_dinamica.head(10).round(2).fillna('-'))
```

### Learning 25: Frequência Cruzada e Agregação (Crosstab)
Tarefa: Use pd.crosstab para descobrir qual categoria de produto gerou o maior volume de vendas em cada Filial.

```python
df['Branch'] = df['Region'].map({'West': 'A', 'East': 'B', 'Central': 'C', 'South': 'D'})

# 2. Cria o Crosstab cruzando Filial x Categoria, somando as Vendas
tabela_cruzada = pd.crosstab(
    index=df['Branch'], 
    columns=df['Category'], 
    values=df['Sales'], 
    aggfunc='sum'
)

# 3. Formata e exibe os resultados para encontrarmos os "melhores"
print("Total de Vendas por Filial e Categoria:\n")
print(tabela_cruzada.round(2))
```

### Learning 28: Regras Tributárias Complexas (Apply com Lambda)
**Tarefa:** Use uma função `lambda` com `.apply()` para recalcular a coluna 'Total' adicionando 5% APENAS se a categoria for 'Electronic accessories'.
